```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Hp\Downloads\heart.csv")

print(df.shape)
print(df.head())
```

```python
df.drop(
    ['oldpeak', 'slope', 'ca', 'thal'],
    axis=1,
    inplace=True
)

print(df.shape)
```

```python
from pgmpy.models import BayesianNetwork

model = BayesianNetwork([
    ('age', 'target'),
    ('sex', 'target'),
    ('cp', 'target'),
    ('trestbps', 'target'),
    ('chol', 'target'),
    ('fbs', 'target'),
    ('restecg', 'target'),
    ('thalach', 'target'),
    ('exang', 'target')
])

print("Bayesian Network created successfully")
```

```python
df_bn = df.copy()

# Convert continuous values into 2 categories
df_bn['age'] = pd.cut(df_bn['age'], bins=2, labels=False)
df_bn['trestbps'] = pd.cut(df_bn['trestbps'], bins=2, labels=False)
df_bn['chol'] = pd.cut(df_bn['chol'], bins=2, labels=False)
df_bn['thalach'] = pd.cut(df_bn['thalach'], bins=2, labels=False)

print(df_bn.head())
```

```python
print("Age:", df_bn['age'].unique())
print("Trestbps:", df_bn['trestbps'].unique())
print("Chol:", df_bn['chol'].unique())
print("Thalach:", df_bn['thalach'].unique())
```

```python
from pgmpy.models import BayesianNetwork

model = BayesianNetwork([
    ('age', 'target'),
    ('sex', 'target'),
    ('cp', 'target'),
    ('trestbps', 'target'),
    ('chol', 'target'),
    ('fbs', 'target'),
    ('restecg', 'target'),
    ('thalach', 'target'),
    ('exang', 'target')
])

print("Bayesian Network created successfully")
```

```python
from pgmpy.estimators import MaximumLikelihoodEstimator

model.fit(
    df_bn,
    estimator=MaximumLikelihoodEstimator
)

print("CPDs learned successfully")
```

```python
for cpd in model.get_cpds():
    print(cpd)
```
